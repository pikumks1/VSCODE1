```python
## Create a simple txt file

import os
os.makedirs("../data/text_files", exist_ok=True)
```

```python
## Create a simple txt file
sample_text = {
"../data/text_files/python_intro.txt": "This is the content of sample1.txt. It contains some example text for testing.",
"../data/text_files/machine_learning.txt": """This is the content of sample2.txt. Machine Learngin contains some example text for testing.""",
}

for file_path, content in sample_text.items():
    with open(file_path, "w", encoding="utf-8") as f:
        f.write(content)

print("Sample text files created successfully.")
```

```python
### Text Loader
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")
documents = loader.load()

print(documents)
```

```python
### Directory loader
from langchain_community.document_loaders import DirectoryLoader

## Load all text files in the directory
dir_loader = DirectoryLoader(
    "../data/text_files", 
    glob="*.txt", #pattern to match only text files
    loader_cls=TextLoader, #specify the loader class to use for each file
    loader_kwargs={"encoding": "utf-8"}, #additional arguments to pass to the loader
    show_progress=False #show progress bar while loading -- need  to import tqdm. so disable it for now
)

documents = dir_loader.load()
print(documents)
```

```python
from langchain_community.document_loaders import PyPDFLoader

dir_loader = DirectoryLoader(
    "../data/pdf",
    glob="*.pdf",
    loader_cls=PyPDFLoader,
    show_progress=False
)

pdf_documents = dir_loader.load()
print(pdf_documents)
```
