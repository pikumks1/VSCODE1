```python
### Test Groq model with langchain-groq

import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq  # Fixed the import here

# 1. Load environment variables from .env file
load_dotenv()

# 2. Get the key from the environment
# Make sure your .env file has: GROQ_API_KEY=your_key_here
groq_api_key = os.getenv("GROQ_API_KEY")

# 3. Initialize the LLM
# Note: If groq_api_key is None, ChatGroq will automatically 
# look for an environment variable named GROQ_API_KEY anyway.
llm = ChatGroq(
    groq_api_key=groq_api_key, 
    model="groq/compound",
    temperature=0.1,
    max_tokens=1024
)

# Optional: Test it
response = llm.invoke("Hello, are you working?")
print(response.content)
```

```python
# Find active models
import requests

url = "https://api.groq.com/openai/v1/models"
headers = {"Authorization": f"Bearer {groq_api_key}"}

response = requests.get(url, headers=headers)
if response.status_code == 200:
    models = response.json()
    for model in models['data']:
        print(model['id'])
else:
    print("Error fetching models. Check your API key.")
```
